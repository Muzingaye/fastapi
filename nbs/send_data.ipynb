```python
base_url = "http://localhost:8000"
path = "/events/"
endpoint =f"{base_url}{path}" 
```

```python
resp = requests.get(endpoint)

if resp.ok:
    data = resp.json()
    pprint(data)
else:
    pprint(resp.text)
```

```python
create_endpoint = f'{base_url}/events'
data = {
    "page":"what page",
    "description": "Test Event Create"
}



res = requests.post(
    create_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```

```python
update_endpoint = f'{base_url}/events/'
data = {
    "page":"what page",
    "description": "Test Event Update"
}

res = requests.put(
    update_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```
