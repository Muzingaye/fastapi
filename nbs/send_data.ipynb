```python
create_endpoint = f'{base_url}/events'
data = {
    "page":"what page",
    "description": "Test Event Create"
}



res = requests.post(
    create_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```

```python
create_endpoint = f'{base_url}/events/14'
data = {
    "page":"what page",
    "description": "Test Event Update"
}

res = requests.put(
    create_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```
