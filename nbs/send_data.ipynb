```python
create_endpoint = f'{base_url}/events'
data = {
    "page":"what page",
    "description": "Test Event Create"
}



res = requests.post(
    create_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```

```python
update_endpoint = f'{base_url}/events/'
data = {
    "page":"what page",
    "description": "Test Event Update"
}

res = requests.put(
    update_endpoint,
    json=data
)
print(res.status_code)
# print(res.headers)
# print(res.json())
if res.ok:
    res.json()
    print(res.status_code)
```

```python
from faker import Faker
import random
events = 100
pages = ["/",'/about', '/contact', '/pages', '/pricing', "/blog", "products", "login", "/dashboard", "/settings"]

fake = Faker()
sess_ids = [fake.uuid4() for i in range(20)]
create_endpoint =f"{base_url}/events"
referrer = [
    "https://google.com", 
    "https://facebook.com",
    "https://linkedin.com",
    "x.com",
    "https://github.com",
    "",
]
# print(random.choices(pages))
for i in range(events):
    payload = dict(
        page= random.choices(pages)[0],
        description =random.choice(pages)[0],
        userAgent =fake.user_agent(),
        ipAddress = fake.ipv4(),
        referrer=random.choice(referrer),
        sessionId = random.choice(sess_ids),
        # duration= random.randint(3, 300)
    )
    response = requests.post(create_endpoint, json=payload)
    if response.ok:
        data = response.json()
        pprint(data)
    else:
        pprint(response.text)
```
